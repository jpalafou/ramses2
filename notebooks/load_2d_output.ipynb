# Load a 2D RAMSES output

```python
from pathlib import Path
import re
import sys
from dataclasses import dataclass, field
import numpy as np

sys.path.insert(0, "../utils/py")
import ramses_io
```

```python
@dataclass
class RamsesOutput:
    x: np.ndarray = field(default_factory=lambda: np.array([]))
    y: np.ndarray = field(default_factory=lambda: np.array([]))
    rho: np.ndarray = field(default_factory=lambda: np.array([]))
    t_finish: float = 0.0

def load_finishing_time(path):
    path = Path(path)
    info = sorted(path.glob("output_*/info.txt"))[-1]
    return ramses_io.rd_info_txt(info)["time"]

def load_snapshot(path, n=2):
    snapshot = path / f"output_{n:05d}"
    cells = ramses_io.leaf_cells(snapshot, fields=["density"])
    x, y = cells.x
    rho = cells["density"]
    t_finish = load_finishing_time(path)
    out = RamsesOutput(x=x, y=y, rho=rho, t_finish=t_finish)
    return out
```

```python
out0 = load_snapshot(Path("../out/sinus2d/old/level6"))
out1 = load_snapshot(Path("../out/sinus2d/1dof/level6"))
```

```python
assert np.array_equal(out0.rho, out1.rho)
```
